## Pandas Dataframes

```python
import pandas as pd
```

```python
data={
    'Name':["Hamim","Fahim","Tasin"],
    'Age':[21,13,15],
    'Class':["1st year","5th","9th"]
}

df = pd.DataFrame(data)
df
```

#### Selection and Indexing Columns

```python
print(df["Age"][0]) # Selectng single column data
print('\n')
print(df[["Name","Class"]])  # Selectng Multiple column data
```

#### Creating New Column

```python
df["Result"]=["A","A+","A-"]
df["Study-hour"]=[2,5,2]

df
```

#### Removing

```python
df.drop("Class",axis=1)
```

#### Selecting rows

```python
print(df.loc[0])
print("\n")
print(df.loc[[0,1]][["Name","Age"]]),,
print("\n")
print(df.iloc[2])
```

#### Conditional Selection

```python
df
```

```python
df[(df["Age"] < 20) & (df["Result"] == "A+") ]
```

### Missing Data

```python

import numpy as np
import pandas as pd


data = {
    "Employee_ID": [
        1001, 1002, 1003, 1004,
        1005, 1006, 1007, 1008,
        1009, 1010, 1011, 1012
    ],

    "Name": [
        "Rahim", "Karim", np.nan, "Sultana",
        "Jamil", np.nan, "Akram", "Fahima",
        "Nabila", "Hasan", np.nan, "Sakib"
    ],

    "Department": [
        "IT", "HR", "Sales", np.nan,
        "IT", "Finance", np.nan, "Marketing",
        "HR", "IT", "Sales", np.nan
    ],

    "Salary": [
        85000, np.nan, 60000, 55000,
        np.nan, 70000, 95000, 62000,
        58000, np.nan, 75000, 68000
    ],

    "Age": [
        25, 30, 28, np.nan,
        35, 32, 27, 29,
        np.nan, 26, 31, 24
    ],

    "Join_Date": [
        "2021-01-15", "2022-03-20", np.nan,
        "2020-11-05", "2023-05-12", np.nan,
        "2019-08-25", "2022-01-10",
        "2021-07-18", "2024-02-01",
        np.nan, "2023-09-15"
    ]
}

df = pd.DataFrame(data)

print(df)
```

```python
df.isna()
```

```python
df.isna().sum()
```

```python
df.isna().any()
```

#### Removing missing data

```python
df.dropna() # removing nan rows
```

```python
df
```

```python
df.dropna(thresh=5)
```

#### Filling Missing Data

```python
df
```

```python
df.fillna(0)
```

```python
salary_value=df["Salary"].mean()
values={"Salary":salary_value}

df.fillna(value=values)
```
