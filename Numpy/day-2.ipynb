## Numpy Aggregation Funtion

```python
import numpy as np
```

```python
arr=np.array([10,20,30,40])


print(arr.sum())
print(arr.mean())
print(arr.max())
print(arr.min())
print(arr.std())
print(arr.var())
```

# Indexing And Slicing

##### Access Element

```python
arr=np.array([1,2,3,4])

print(arr[0])  # First element of array
print(arr[-1]) # Last elenment of array
```

##### MultiDimentional

```python
arr=np.array(
              [
                  [1,2,3],
                  [4,5,6]
              ]
             )


# arr[row,column]

print(arr[0]) # First row 
print(arr[1]) # Second row
print(arr[0,0]) # first row first column element
print(arr[0,1]) # first row second column element
print(arr[0,2]) # first row third column element


```

##### Slicing Elements

<p><b>---> arr[start:end:step]</b></p>

```python
arr=np.array([10,20,30,40,50,60])

print(arr[1:3]) # 1 to 2 index
print(arr[:4]) # 0 to 3 index
print(arr[::2]) # index (0,2,4)
print(arr[::-1]) # Reverse 
```

##### Fancy Indexing

```python
arr=np.array([10,20,30,40,50,60])

arr[[0,2,3]]
```

#### Filtering {Boolean masking}

```python
arr=np.array([10,20,30,40,50,60])


arr[arr < 48]
```
