# Numpy Tutorial start

```python
import numpy as np
```

## Numpy Array Properties

### Shape

```python
arr=np.array([[1,2,3],
              [4,5,6]]) # 2 rows, 3 columns 


arr.shape
```

### Size (Total number of element)

```python
arr=np.array([[1,2,3],
              [4,5,6]]) # 6 elements in this 2D array


print(f'Size of this array = {arr.size}')
```

### Array Dimension

```python
array_1d=np.array([1,2,3,4])
array_2d=np.array([[1,2,3],[4,5,5],[10,22,3]])
array_3d=np.array([[[1,2,3],[22,44,10]]])



print(array_1d.ndim)
print(array_2d.ndim)
print(array_3d.ndim)
```

### Data type

```python
arr=np.array([1,2,3,2.44])

arr.dtype
```

## Numpy Method

#### Changing Data Types

```python
arr=np.array([1.22,3.33,22,8.99])
new_arr=arr.astype(int)

print(new_arr)
print(new_arr.dtype)
```

### Array Math

```python
arr=np.array([1,2,3])

print(arr + 2)
print(arr - 0.5)
print(arr * 5)
print(arr / 2)
```
